```python
# ==========================================================
# SISTEM FUZZY PENGATURAN SUHU RUANG SERVER
# Metode: Mamdani
# Input : Suhu dan Kelembapan
# Output: Tingkat Pendinginan
# ==========================================================

!pip install -q scikit-fuzzy

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import skfuzzy as fuzz
from skfuzzy import control as ctrl


# ==========================================================
# 1. VARIABEL FUZZY
# ==========================================================

suhu = ctrl.Antecedent(
    np.arange(15, 36, 0.1),
    'suhu'
)

kelembapan = ctrl.Antecedent(
    np.arange(20, 91, 0.1),
    'kelembapan'
)

pendinginan = ctrl.Consequent(
    np.arange(0, 101, 0.1),
    'pendinginan'
)


# ==========================================================
# 2. MEMBERSHIP FUNCTION SUHU
# ==========================================================

suhu['dingin'] = fuzz.trapmf(
    suhu.universe,
    [15, 15, 20, 24]
)

suhu['normal'] = fuzz.trimf(
    suhu.universe,
    [22, 25, 28]
)

suhu['panas'] = fuzz.trapmf(
    suhu.universe,
    [26, 29, 35, 35]
)


# ==========================================================
# 3. MEMBERSHIP FUNCTION KELEMBAPAN
# ==========================================================

kelembapan['kering'] = fuzz.trapmf(
    kelembapan.universe,
    [20, 20, 35, 50]
)

kelembapan['sedang'] = fuzz.trimf(
    kelembapan.universe,
    [40, 55, 70]
)

kelembapan['lembap'] = fuzz.trapmf(
    kelembapan.universe,
    [60, 75, 90, 90]
)


# ==========================================================
# 4. MEMBERSHIP FUNCTION OUTPUT
# ==========================================================

pendinginan['rendah'] = fuzz.trapmf(
    pendinginan.universe,
    [0, 0, 20, 40]
)

pendinginan['sedang'] = fuzz.trimf(
    pendinginan.universe,
    [30, 50, 70]
)

pendinginan['tinggi'] = fuzz.trapmf(
    pendinginan.universe,
    [60, 80, 100, 100]
)


# ==========================================================
# 5. RULE FUZZY
# ==========================================================

rules = [

    ctrl.Rule(
        suhu['dingin'] & kelembapan['kering'],
        pendinginan['rendah']
    ),

    ctrl.Rule(
        suhu['dingin'] & kelembapan['sedang'],
        pendinginan['rendah']
    ),

    ctrl.Rule(
        suhu['dingin'] & kelembapan['lembap'],
        pendinginan['sedang']
    ),

    ctrl.Rule(
        suhu['normal'] & kelembapan['kering'],
        pendinginan['rendah']
    ),

    ctrl.Rule(
        suhu['normal'] & kelembapan['sedang'],
        pendinginan['sedang']
    ),

    ctrl.Rule(
        suhu['normal'] & kelembapan['lembap'],
        pendinginan['tinggi']
    ),

    ctrl.Rule(
        suhu['panas'] & kelembapan['kering'],
        pendinginan['tinggi']
    ),

    ctrl.Rule(
        suhu['panas'] & kelembapan['sedang'],
        pendinginan['tinggi']
    ),

    ctrl.Rule(
        suhu['panas'] & kelembapan['lembap'],
        pendinginan['tinggi']
    )
]


# ==========================================================
# 6. MEMBUAT CONTROL SYSTEM
# ==========================================================

sistem = ctrl.ControlSystem(rules)


# ==========================================================
# 7. FUNGSI PREDIKSI
# ==========================================================

def prediksi(suhu_input, rh_input):

    simulasi = ctrl.ControlSystemSimulation(sistem)

    simulasi.input['suhu'] = suhu_input
    simulasi.input['kelembapan'] = rh_input

    simulasi.compute()

    return simulasi.output['pendinginan']


# ==========================================================
# 8. KATEGORI OUTPUT
# ==========================================================

def kategori(nilai):

    if nilai < 40:
        return "Rendah"

    elif nilai < 70:
        return "Sedang"

    else:
        return "Tinggi"


# ==========================================================
# 9. CONTOH SIMULASI
# ==========================================================

suhu_input = 28
rh_input = 65

hasil = prediksi(
    suhu_input,
    rh_input
)

print("==========================================")
print("HASIL SIMULASI SISTEM FUZZY")
print("==========================================")
print(f"Suhu          : {suhu_input} °C")
print(f"Kelembapan    : {rh_input} %")
print(f"Pendinginan   : {hasil:.2f} %")
print(f"Kategori      : {kategori(hasil)}")
print("==========================================")


# ==========================================================
# 10. GRAFIK MEMBERSHIP FUNCTION
# ==========================================================

suhu.view()
plt.title("Membership Function Suhu")
plt.show()

kelembapan.view()
plt.title("Membership Function Kelembapan")
plt.show()

pendinginan.view()
plt.title("Membership Function Pendinginan")
plt.show()


# ==========================================================
# 11. PENGUJIAN BEBERAPA KONDISI
# ==========================================================

data = [
    [20, 35],
    [23, 45],
    [25, 55],
    [27, 60],
    [28, 65],
    [30, 70],
    [32, 80],
    [34, 85]
]

hasil_data = []

for t, rh in data:

    output = prediksi(t, rh)

    hasil_data.append([
        t,
        rh,
        round(output, 2),
        kategori(output)
    ])


df = pd.DataFrame(
    hasil_data,
    columns=[
        "Suhu (°C)",
        "Kelembapan (%)",
        "Pendinginan (%)",
        "Kategori"
    ]
)

print("\nHASIL PENGUJIAN")
display(df)


# ==========================================================
# 12. GRAFIK HASIL PENGUJIAN
# ==========================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(len(df)),
    df["Pendinginan (%)"],
    marker="o"
)

plt.xlabel("Skenario Pengujian")
plt.ylabel("Pendinginan (%)")
plt.title("Hasil Pengujian Sistem Fuzzy")

plt.grid(True)

plt.show()
```
